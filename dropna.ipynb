```python
df = spark.read.format("csv").option("header", "True").option("inferSchema", "True").load("/Volumes/workspace/default/workout_practise/Input/Customers_Workout2.csv")
df.display()
```

```python
# #when(condition,value).otherwise(),  fill value with isNull() use None.
from pyspark.sql.functions import when, col

df1 = df.withColumn("EmailAddress", when(col("EmailAddress") == "shannon38@adventure-works.com", None).otherwise(col("EmailAddress")))\
        .withColumn("FirstName", when (col("FirstName") == "LAUREN",None).otherwise(col("FirstName")))\
        .withColumn("TotalChildren", when (col("TotalChildren") == 2,None).otherwise(col("TotalChildren")))
df1.display()
```

```python
# Compute count of nulls for FirstName, EmailAddress, and TotalChildren columns
from pyspark.sql.functions import when, col, sum
second = df1.select(
    sum(when(col("FirstName").isNull(), 1).otherwise(0)).alias("Namecount"),
    sum(when(col("EmailAddress").isNull(), 1).otherwise(0)).alias("emailcount"),
    sum(when(col("TotalChildren").isNull(), 1).otherwise(0)).alias("TotalChilcount")
)
second.display()
```
