```python
dfgoldraw=spark.read.format("delta").table("deltacatalog.deltaschema.silver_sales")
dfgoldraw.display()
```

```python
from pyspark.sql.functions import count,col, sum as _sum
dfgoldagg=dfgoldraw.groupBy("date").agg(
    count(col("*")).alias("count"),
    sum(col("OrderQuantity")).alias("sum_Quantity")
      )
dfgoldagg.display()
```

```python
dfgoldagg.write.mode("overwrite").format("delta").saveAsTable("deltacatalog.deltaschema.gold_sales")
```

```python
dfgoldraw = spark.read.format("delta").table("deltacatalog.deltaschema.silver_sales")

from pyspark.sql.functions import count, col, sum as _sum

dfgoldagg = dfgoldraw.groupBy("date").agg(
    count("*").alias("total_records"),
    count("OrderNumber").alias("total_orders"),
    _sum("OrderQuantity").alias("total_quantity")
)

dfgoldagg.display()

dfgoldagg.write.mode("overwrite") \
    .format("delta") \
    .saveAsTable("deltacatalog.deltaschema.gold_sales")
```
