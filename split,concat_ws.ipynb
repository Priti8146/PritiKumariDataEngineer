Sting to list and list to string

```python
column=["comment","name","age"]
data=[("i like you","Priti",29),("I love you","Puja",32),("I miss you","kumari",20)]
df=spark.createDataFrame(data,column)
df.display()
```

```python
from pyspark.sql.functions import col,split
df1=df.withColumn("comment_list",split(col("comment")," "))
df1.display()
```

```python
df1.printSchema()
```

```python
from pyspark.sql.functions import col,split,concat_ws
df2=df1.withColumn("comment_string",concat_ws(" ", col("comment_list")))
df2.display()
```

```python
df2.printSchema()
```
