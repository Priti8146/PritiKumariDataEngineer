```python
# Minimal fix of list and DataFrame creation
# Define the data as a list of dictionaries

df = [
    {"Name": "Priti", "Age": 20, "height": 5},
    {"Name": "Puja", "Age": 21, "height": 5},
    {"Name": "Priti", "Age": 20, "height": 6},
    {"Name": "Puja", "Age": 21, "height": 5}
]
df = spark.createDataFrame(df)
df.display()
```

```python
# Drop exact duplicate rows
df1 = df.dropDuplicates().display()
```

```python
# Drop duplicates based on specific columns
df2 = df.dropDuplicates(["height"]).display()
```
