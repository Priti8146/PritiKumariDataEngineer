```python
data= [
      {"comment": ["I","love", "you"], "name" : "Priti", "age": 30},
      {"comment": ["I","like", "you"], "name" : "Puja", "age":20},
      {"comment": ["I","miss", "you"], "name" :"Kumari", "age":40}    
      ]
df = spark.createDataFrame(data)
df.display()
```

```python
df2=df.withColumn("commentnew", explode(col("comment")))
df2.display()
```

```python
from pyspark.sql.functions import collect_list
df3=df2.groupBy(col("commentnew")).agg(collect_list(col("commentnew")).alias("collectlist"))
df3.display()
```

```python
from pyspark.sql import Window
from pyspark.sql.functions import collect_list
df5 = df2.withColumn("list", collect_list(col("commentnew")).over(Window.partitionBy(col("commentnew"))))
df5.display()
```
