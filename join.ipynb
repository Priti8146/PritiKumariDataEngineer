```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("JoinExample").getOrCreate()

# Table A
dataA = [(1,), (1,), (1,), (1,)]
columns = ["col1"]
dfA = spark.createDataFrame(dataA, columns)

# Table B
dataB = [(1,), (1,), (None,), (None,)]
dfB = spark.createDataFrame(dataB, columns)

dfA.show()
dfB.show()
```

```python
inner_df = dfA.join(dfB, dfA.col1 == dfB.col1, "inner")
inner_df.display()
```

```python
inner_df = dfA.join(dfB, dfA.col1 == dfB.col1, "left")
inner_df.display()
```

```python
inner_df = dfA.join(dfB, dfA.col1 == dfB.col1, "right")
inner_df.display()
```

```python
inner_df = dfA.join(dfB, dfA.col1 == dfB.col1, "anti")
inner_df.display()
```
