```python
colu = ["dep","date","sales","Price"]
data=[
    ("IT","23/02/2026","Bag",100),
    ("IT","24/02/2026","Bag",200),
    ("CSC","23/02/2026","Bag",100),
    ("CSC","23/02/2026","pen",100),
    ("CSC","24/02/2026","boll",80),
    ("CSC","25/02/2026","Bag",1500),
    ("CSC","28/02/2026","Bag",100),
    ("Finance","23/02/2026","Bag",100),
    ("", "23/02/2026","Bag",100)
]
df=spark.createDataFrame(data, colu)
df.display()
```

```python
from pyspark.sql.functions import to_date
dfdate=df.withColumn("date",to_date("date", "dd/MM/yyyy"))
dfdate.display()
dfdate.printSchema()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lead,col

windowspec=Window.partitionBy("dep").orderBy("date")

df1=dfdate.withColumn("lead1", lead("Price",1).over(windowspec)) \
          .filter(col("dep") != "")
df1.display()
```

```python
from pyspark.sql.functions import substr,col
df1price=df1.withColumn("pricediff",col("lead1")-col("Price"))
          
df1price.display()
```

```python
from pyspark.sql.functions import coalesce,lit,col
df_final = df1.withColumn("price_diff1", coalesce(col("lead1"), lit(0)) - col("price")).display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import lead,lag,datediff

windowspec=Window.partitionBy("dep").orderBy("date")

df4=dfdate.withColumn(("leaddate"), lead("date",1).over(windowspec))
df4.display()
```

```python
dfdiff=df4.withColumn(("datediff"),datediff(col("leaddate"),col("date")))
dfdiff.display()
```
